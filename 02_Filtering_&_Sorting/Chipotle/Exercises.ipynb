```python
chipo
```

### Step 4. How many products cost more than $10.00?

```python
chipo['item_price'] = chipo['item_price'].apply(lambda x:float(x.replace('$','')))


```

```python
chipo['individual_price'] = chipo['item_price']/chipo['quantity']
```

```python
price_df = chipo[['item_name','choice_description','individual_price']].drop_duplicates(subset=['item_name','choice_description'])
```

```python
price_df['individual_price'].apply(lambda x: x > 10.00).value_counts()[True]
```
